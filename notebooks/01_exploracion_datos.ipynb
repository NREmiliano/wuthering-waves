```python
import os

ruta_bindata = "../WutheringWaves_Data/BinData"
palabras_clave = ["role", "character", "weapon", "card", "avatar"]

if os.path.exists(ruta_bindata):
  archivos = os.listdir(ruta_bindata)
  print(f"Total de archivos en BinData: {len(archivos)}\n")

  print("--- Archivos relevantes encontrados ---")
  for archivo in archivos:
    if any(palabra in archivo.lower() for palabra in palabras_clave):
      print(f" - {archivo}")
else:
  print("La ruta de BinData no es correcta.")
```

```python
import json
import os

ruta_carpeta_role = "../WutheringWaves_Data/BinData/role"
ruta_roleinfo = os.path.join(ruta_carpeta_role, "roleinfo.json")

if os.path.exists(ruta_roleinfo):
  with open(ruta_roleinfo, "r", encoding="utf-8") as f:
    datos = json.load(f)
    print(f"Estructura cargada exitosamente. Tipo: {type(datos)}")

    # Mostramos los primeros datos formateados
    if isinstance(datos, list) and len(datos) > 0:
      print("\nPrimer personaje registrado:")
      display(datos[0])
    elif isinstance(datos, dict):
      primera_clave = list(datos.keys())[0]
      print(f"\nDatos para la clave {primera_clave}:")
      display(datos[primera_clave])
else:
  print("Revisa la ruta del archivo roleinfo.json")
```
