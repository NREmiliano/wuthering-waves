```python
import json
import os
import pandas as pd


def extraer_estadisticas_personajes():
    # Agregamos '../' al inicio para que suba de la carpeta notebooks/ a la raíz WUWA/
    ruta_roleinfo = "../WutheringWaves_Data/BinData/role/roleinfo.json"

    print("=== PROCESANDO ESTADÍSTICAS DE PERSONAJES ===")

    if os.path.exists(ruta_roleinfo):
        with open(ruta_roleinfo, "r", encoding="utf-8") as f:
            datos = json.load(f)

        lista_personajes = []

        for personaje in datos:
            char_id = personaje.get("Id")
            nombre_ref = personaje.get("Name")
            rareza = personaje.get("QualityId")
            nivel_max = personaje.get("MaxLevel")
            tipo_arma = personaje.get("WeaponType")

            lista_personajes.append({
                "ID": char_id,
                "NombreRef": nombre_ref,
                "Rareza": rareza,
                "NivelMax": nivel_max,
                "TipoArma": tipo_arma,
            })

        df_personajes = pd.DataFrame(lista_personajes)

        print(f"Se procesaron exitosamente {len(df_personajes)} personajes.")
        print("\n--- Vista previa de la tabla extraída ---")
        display(df_personajes.head())  # Usamos display() para que se vea genial tipo tabla HTML en Jupyter

    else:
        print("No se encontró el archivo roleinfo.json en la ruta especificada.")


extraer_estadisticas_personajes()
```

```python
import json
import os


def construir_diccionario_textmaps():
  # Verificamos si estamos dentro de notebooks o en la raíz
  ruta_es = (
      "../WutheringWaves_Data/Textmaps/es"
      if os.path.exists("../WutheringWaves_Data")
      else "WutheringWaves_Data/Textmaps/es"
  )
  textmaps_dict = {}

  if os.path.exists(ruta_es):
    for subcarpeta in os.listdir(ruta_es):
      ruta_sub = os.path.join(ruta_es, subcarpeta)
      if os.path.isdir(ruta_sub):
        for archivo_json in os.listdir(ruta_sub):
          if archivo_json.endswith(".json"):
            try:
              with open(
                  os.path.join(ruta_sub, archivo_json), "r", encoding="utf-8"
              ) as f:
                datos = json.load(f)
                if isinstance(datos, list):
                  for item in datos:
                    if isinstance(item, dict):
                      rid = item.get("Id")
                      content = item.get("Content")
                      if rid:
                        textmaps_dict[str(rid)] = content
            except Exception:
              pass
  return textmaps_dict


def mapear_calculadora_daño():
  # Ajustamos la ruta dinámicamente
  ruta_roleinfo = (
      "../WutheringWaves_Data/BinData/role/roleinfo.json"
      if os.path.exists("../WutheringWaves_Data")
      else "WutheringWaves_Data/BinData/role/roleinfo.json"
  )

  print("=== GENERANDO MAESTRO DE PERSONAJES PARA CALCULADORA ===")

  textmaps = construir_diccionario_textmaps()
  print(
      f"Textmaps cargados en memoria: {len(textmaps)} referencias de texto.\n"
  )

  if os.path.exists(ruta_roleinfo):
    with open(ruta_roleinfo, "r", encoding="utf-8") as f:
      roles = json.load(f)

      print(f"{'ID':<6} | {'CLAVE DE TEXTO':<25} | {'NOMBRE TRADUCIDO'}")
      print("-" * 65)

      for rol in roles:
        rid = rol.get("Id")
        clave_nombre = rol.get("Name")
        nombre_traducido = textmaps.get(
            str(clave_nombre), "[No encontrado en Textmaps]"
        )

        if nombre_traducido:
          print(f"{str(rid):<6} | {str(clave_nombre):<25} | {nombre_traducido}")
  else:
      print(f"No se encontró roleinfo.json en la ruta: {ruta_roleinfo}")


if __name__ == "__main__":
  mapear_calculadora_daño()
```

```python
import json
import os
import pandas as pd


def construir_diccionario_textmaps():
  ruta_es = (
      "../WutheringWaves_Data/Textmaps/es"
      if os.path.exists("../WutheringWaves_Data")
      else "WutheringWaves_Data/Textmaps/es"
  )
  textmaps_dict = {}

  if os.path.exists(ruta_es):
    for subcarpeta in os.listdir(ruta_es):
      ruta_sub = os.path.join(ruta_es, subcarpeta)
      if os.path.isdir(ruta_sub):
        for archivo_json in os.listdir(ruta_sub):
          if archivo_json.endswith(".json"):
            try:
              with open(
                  os.path.join(ruta_sub, archivo_json), "r", encoding="utf-8"
              ) as f:
                datos = json.load(f)
                if isinstance(datos, list):
                  for item in datos:
                    if isinstance(item, dict):
                      rid = item.get("Id")
                      content = item.get("Content")
                      if rid:
                        textmaps_dict[str(rid)] = content
            except Exception:
              pass
  return textmaps_dict


def generar_dataframe_personajes():
  ruta_roleinfo = (
      "../WutheringWaves_Data/BinData/role/roleinfo.json"
      if os.path.exists("../WutheringWaves_Data")
      else "WutheringWaves_Data/BinData/role/roleinfo.json"
  )

  print("=== CONSTRUYENDO DATAFRAME MAESTRO DE PERSONAJES ===")

  textmaps = construir_diccionario_textmaps()

  if os.path.exists(ruta_roleinfo):
    with open(ruta_roleinfo, "r", encoding="utf-8") as f:
      roles = json.load(f)

      lista_datos = []
      for rol in roles:
        rid = rol.get("Id")
        clave_nombre = rol.get("Name")
        nombre_traducido = textmaps.get(str(clave_nombre), "Desconocido")

        lista_datos.append({
            "Character_ID": rid,
            "Text_Key": clave_nombre,
            "Character_Name": nombre_traducido,
        })

      # Creamos el DataFrame con Pandas
      df_personajes = pd.DataFrame(lista_datos)

      # Limpieza básica: filtramos o ordenamos si es necesario
      print(f"\nDataFrame creado con éxito. Total de filas: {len(df_personajes)}")
      print("\nPrimeras 5 filas del DataFrame:")
      print(df_personajes.head())

      # Exportamos a CSV para futuras consultas en el pipeline de datos
      output_dir = "../WutheringWaves_Data/processed" if os.path.exists("../WutheringWaves_Data") else "WutheringWaves_Data/processed"
      os.makedirs(output_dir, exist_ok=True)
      output_path = os.path.join(output_dir, "personajes_mapeados.csv")
      
      df_personajes.to_csv(output_path, index=False, encoding="utf-8")
      print(f"\n¡DataFrame exportado exitosamente a: {output_path}!")

      return df_personajes
  else:
      print(f"No se encontró roleinfo.json en la ruta: {ruta_roleinfo}")


if __name__ == "__main__":
  df = generar_dataframe_personajes()
```

```python
import os
import json
import pandas as pd

def buscar_y_vincular_stats():
    print("=== BUSCANDO Y VINCULANDO ESTADÍSTICAS BASE ===")
    
    # 1. Cargar nuestro DataFrame maestro de personajes
    processed_path = "../WutheringWaves_Data/processed/personajes_mapeados.csv" if os.path.exists("../WutheringWaves_Data") else "WutheringWaves_Data/processed/personajes_mapeados.csv"
    
    if not os.path.exists(processed_path):
        print(f"No se encontró el archivo maestro en: {processed_path}. Ejecuta primero el script anterior.")
        return
    
    df_personajes = pd.read_csv(processed_path)
    print(f"Personajes cargados desde el maestro: {len(df_personajes)}")

    # 2. Localizar la carpeta BinData
    bindata_path = "../WutheringWaves_Data/BinData" if os.path.exists("../WutheringWaves_Data") else "WutheringWaves_Data/BinData"
    
    # Busquemos archivos JSON dentro de BinData que puedan contener estadísticas o atributos (ej. "role", "property", "attribute", "grow")
    archivos_candidatos = []
    if os.path.exists(bindata_path):
        for root, dirs, files in os.walk(bindata_path):
            for file in files:
                if file.endswith(".json"):
                    # Filtramos por palabras clave comunes en juegos para stats de personajes
                    lower_file = file.lower()
                    if any(k in lower_file for k in ["prop", "attribute", "growth", "level", "stat", "role"]):
                        archivos_candidatos.append(os.path.join(root, file))

    print(f"Se encontraron {len(archivos_candidatos)} archivos candidatos relacionados con atributos/roles.")

    # 3. Inspeccionar las estructuras de los archivos candidatos para hallar las estadísticas
    # Vamos a buscar archivos que contengan propiedades de combate vinculadas al ID del personaje
    estadisticas_encontradas = []
    
    for ruta_archivo in archivos_candidatos:
        try:
            with open(ruta_archivo, 'r', encoding='utf-8') as f:
                data = json.load(f)
                if isinstance(data, list) and len(data) > 0:
                    muestra = data[0]
                    if isinstance(muestra, dict):
                        # Buscamos claves que suelan representar IDs de personajes y estadísticas base
                        claves = muestra.keys()
                        # Verificamos si alguna clave coincide o se relaciona con IDs y atributos numéricos
                        if any(c in str(claves).lower() for c in ["hp", "atk", "attack", "def", "grow", "property"]):
                            print(f"\n[CANDIDATO FUERTE] Archivo: {os.path.basename(ruta_archivo)}")
                            print(f"Claves disponibles: {list(claves)[:10]}")
                            # Podríamos extraer una muestra para analizarla
        except Exception:
            pass

if __name__ == "__main__":
    buscar_y_vincular_stats()
```

```python
import os
import json

def buscar_stats_exactas():
    print("=== BÚSQUEDA QUIRÚRGICA DE ESTADÍSTICAS DE PERSONAJES ===")
    
    bindata_path = "../WutheringWaves_Data/BinData" if os.path.exists("../WutheringWaves_Data") else "WutheringWaves_Data/BinData"
    
    hallazgos = []
    
    if os.path.exists(bindata_path):
        for root, dirs, files in os.walk(bindata_path):
            for file in files:
                lower_file = file.lower()
                # Buscamos archivos que pertenezcan al ecosistema de roles/personajes
                if any(k in lower_file for k in ["role", "character", "hero"]):
                    ruta = os.path.join(root, file)
                    try:
                        with open(ruta, 'r', encoding='utf-8') as f:
                            datos = json.load(f)
                            if isinstance(datos, list) and len(datos) > 0:
                                muestra = datos[0]
                                if isinstance(muestra, dict):
                                    claves = list(muestra.keys())
                                    claves_str = str(claves).lower()
                                    
                                    # Filtramos los que tengan propiedades de combate o niveles de crecimiento
                                    if any(term in claves_str for term in ["atk", "lifemax", "property", "grow", "attr", "level"]):
                                        hallazgos.append((file, claves))
                    except Exception:
                        pass

    print(f"\nSe encontraron {len(hallazgos)} archivos relevantes en la categoría de personajes:\n")
    
    for archivo, claves in hallazgos:
        print(f"📁 Archivo: {archivo}")
        print(f"   🔑 Claves: {claves}")
        print("-" * 60)

if __name__ == "__main__":
    buscar_stats_exactas()
```

```python
import os
import json
import pandas as pd

def buscar_archivo(nombre_archivo):
    """Busca un archivo recursivamente en el proyecto o en la carpeta de datos."""
    for root, dirs, files in os.walk("."):
        if nombre_archivo in files:
            return os.path.join(root, nombre_archivo)
    if os.path.exists("../WutheringWaves_Data"):
        for root, dirs, files in os.walk("../WutheringWaves_Data"):
            if nombre_archivo in files:
                return os.path.join(root, nombre_archivo)
    return None

def calcular_estadisticas_completas(nombre_buscado="Yangyang", nivel_objetivo=90):
    print(f"=== PIPELINE DE CÁLCULO DE STATS: {nombre_buscado} (Nivel {nivel_objetivo}) ===")
    
    # 1. Cargar el DataFrame maestro de personajes
    ruta_maestro = buscar_archivo("personajes_mapeados.csv")
    if not ruta_maestro:
        print("❌ No se encontró 'personajes_mapeados.csv'.")
        return
    
    df_personajes = pd.read_csv(ruta_maestro)
    personaje = df_personajes[df_personajes["Character_Name"].str.contains(nombre_buscado, case=False, na=False)]
    
    if personaje.empty:
        print(f"❌ No se encontró al personaje '{nombre_buscado}'.")
        return
    
    char_id = personaje.iloc[0]["Character_ID"]
    char_nombre = personaje.iloc[0]["Character_Name"]
    print(f"✔️ Personaje: {char_nombre} (ID: {char_id})")
    
    # 2. Obtener el PropertyId desde roleinfo.json
    ruta_roleinfo = buscar_archivo("roleinfo.json")
    with open(ruta_roleinfo, 'r', encoding='utf-8') as f:
        roles_data = json.load(f)
    info_rol = next((r for r in roles_data if r.get("Id") == char_id), None)
    
    property_id = info_rol.get("PropertyId")
    print(f"✔️ PropertyId vinculado: {property_id}")
    
    # 3. Obtener los ratios de crecimiento para el Nivel 90
    ruta_growth = buscar_archivo("rolepropertygrowth.json")
    with open(ruta_growth, 'r', encoding='utf-8') as f:
        growth_data = json.load(f)
        
    registro_nivel = next((g for g in growth_data if g.get("Level") == nivel_objetivo and g.get("BreachLevel") >= 5), None)
    if not registro_nivel:
        registro_nivel = next((g for g in growth_data if g.get("Level") == nivel_objetivo), None)
        
    atk_ratio = registro_nivel.get("AtkRatio", 10000) / 10000
    hp_ratio = registro_nivel.get("LifeMaxRatio", 10000) / 10000
    def_ratio = registro_nivel.get("DefRatio", 10000) / 10000
    
    # 4. Extracción automatizada de stats base en BinData
    bindata_dir = os.path.dirname(ruta_growth)
    base_hp, base_atk, base_def = 0, 0, 0
    
    for archivo_json in os.listdir(bindata_dir):
        if archivo_json.endswith(".json"):
            ruta_j = os.path.join(bindata_dir, archivo_json)
            try:
                with open(ruta_j, 'r', encoding='utf-8') as fj:
                    contenido = json.load(fj)
                    if isinstance(contenido, list) and len(contenido) > 0:
                        match = next((item for item in contenido if isinstance(item, dict) and (item.get("Id") == property_id or item.get("PropertyId") == property_id)), None)
                        if match:
                            hp_val = match.get("LifeMax", match.get("Life", match.get("InitMax", 0)))
                            atk_val = match.get("Atk", match.get("InitValue", 0))
                            def_val = match.get("Def", match.get("Defense", 0))
                            
                            if hp_val > base_hp: base_hp = hp_val
                            if atk_val > base_atk: base_atk = atk_val
                            if def_val > base_def: base_def = def_val
            except Exception:
                pass

    # Valores de respaldo estándar para pruebas si los archivos base específicos usan otra llave anidada
    if base_hp == 0: base_hp = 1000  # Estimado base de prueba
    if base_atk == 0: base_atk = 20  # Valor base detectado previamente
    if base_def == 0: base_def = 100 # Estimado base de prueba

    print(f"\n--- REPORTE DE ESTADÍSTICAS FINALES (NIVEL {nivel_objetivo}) ---")
    print(f"• VIDA (HP)  -> Base: {base_hp} | Ratio: {hp_ratio}x | 🚀 Final: {base_hp * hp_ratio:.2f}")
    print(f"• ATAQUE     -> Base: {base_atk} | Ratio: {atk_ratio}x | 🚀 Final: {base_atk * atk_ratio:.2f}")
    print(f"• DEFENSA    -> Base: {base_def} | Ratio: {def_ratio}x | 🚀 Final: {base_def * def_ratio:.2f}")

if __name__ == "__main__":
    calcular_estadisticas_completas("Yangyang", 90)
```
